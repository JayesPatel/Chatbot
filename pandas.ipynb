```python
# Chatbot
import json
import random
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import TfidfVectorizer
import joblib

with open("data2.json","r") as file:
    data=json.load(file)
# print(data) 
# Arange Data
# Patten = [hii]
# tag = [g]
# res = { tag : response}
sentence=[] # patterns
labels=[] # tag
response={} # tag : response

for i in data["intents"]:
    tag = i['tag']
    response[tag]=i["responses"]
    for patten in i['patterns']:
     sentence.append(patten)
     labels.append(tag)
print(sentence) 
print()
print(labels)
print()
print(response)
vector = TfidfVectorizer()
x=vector.fit_transform(sentence)
model=MultinomialNB()
model.fit(x, labels)
joblib.dump(model,"chatbot.pkl")
joblib.dump(vector,"vector.pkl")
joblib.dump(response,"res.pkl")
```

```python
import joblib
import random
model = joblib.load("chatbot.pkl")
vector=joblib.load("vector.pkl")
res=joblib.load("res.pkl")
while True:
    user=input("You : ")
    if user=="exit":
        print("Bot : Thanks")
        break
    user_vector=vector.transform([user]) 
    predict=model.predict(user_vector.toarray())[0]
    if predict in res:
        reply = random.choice(res[predict])
        print("Bot : ",reply)
```
